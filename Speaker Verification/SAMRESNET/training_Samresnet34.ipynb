# 2. Augmentation

```python
#Cell 2
class SoftAugmentor:
    def __init__(self, noise_dir=None, sr=16000):
        self.sr = sr
        self.noise_dir = noise_dir
        self.noise_files = glob.glob(os.path.join(noise_dir, '**/*.wav'), recursive=True) if noise_dir else []
        print(f"Đã load {len(self.noise_files)} file MUSAN noise")

    def _normalize(self, wav):
        return wav / (np.max(np.abs(wav)) + 1e-9)
    
    def add_musan_noise(self, wav):
        if not self.noise_files: return wav
        import librosa
        noise_path = np.random.choice(self.noise_files)
        noise, _ = librosa.load(noise_path, sr=self.sr)
        
        if len(noise) < len(wav):
            tile_factor = int(np.ceil(len(wav) / len(noise)))
            noise = np.tile(noise, tile_factor)
        
        start = np.random.randint(0, max(1, len(noise) - len(wav)))
        noise = noise[start:start + len(wav)]
        
        snr_db = np.random.uniform(10, 20)
        signal_power = np.mean(wav ** 2) + 1e-10
        noise_power = np.mean(noise ** 2) + 1e-10
        scale = np.sqrt(signal_power / (10 ** (snr_db / 10) * noise_power))
    
        noisy = wav + scale * noise
        return self._normalize(noisy)

    def change_device(self, wav):
        device = np.random.choice(['mobile', 'cheap_mic'])
        if device == 'mobile':
            b, a = sps.butter(4, [300/(0.5*self.sr), 3400/(0.5*self.sr)], btype='band')
            wav_back = sps.lfilter(b, a, wav)
            wav_back += np.random.normal(0, 0.002, len(wav_back)) 
        else:
            hp = np.random.uniform(150, 300)
            lp = np.random.uniform(3500, 5000)
            b, a = sps.butter(4, [hp/(0.5*self.sr), lp/(0.5*self.sr)], btype='band')
            wav_back = sps.lfilter(b, a, wav)
            wav_back = np.tanh(wav_back * np.random.uniform(1.5, 2.5))
            wav_back += np.random.normal(0, 0.001, len(wav_back))
        return self._normalize(wav_back)

    def apply_augmentation(self, wav, aug_type):
        aug_map = {
            'musan_noise': self.add_musan_noise,
            'device': self.change_device
        }
        # Nếu aug_type không có trong dictionary (ví dụ 'rir_reverb'), trả về nguyên bản wav
        return aug_map.get(aug_type, lambda x: x)(wav)
```

## Load model

```python
#Cell 3
from cli.hub import Hub
from models.speaker_model import get_speaker_model

def load_model_pt(model_dir):
    if model_dir in Hub.Assets:
        model_dir = Hub.get_model(model_dir)
            
    with open(os.path.join(model_dir, 'config.yaml'), 'r') as f:
        config = yaml.load(f, Loader=yaml.FullLoader)
        
    model = get_speaker_model(config['model'])(**config['model_args'])
    ckpt_path = os.path.join(model_dir, 'avg_model.pt')
    checkpoint = torch.load(ckpt_path, map_location='cpu')
    state_dict = checkpoint['state_dict'] if 'state_dict' in checkpoint else checkpoint
    
    model.load_state_dict(state_dict, strict=False)
    return model, config

def compute_fbank(waveform, sample_rate=16000, num_mel_bins=80, frame_length=25, frame_shift=10):
    feat = kaldi.fbank(waveform,
        num_mel_bins=num_mel_bins, frame_length=frame_length,
        frame_shift=frame_shift, sample_frequency=sample_rate, window_type='hamming')
    feat = feat - torch.mean(feat, 0)
    return feat
```

# 3. Load train, valid dataset


```python
#Cell 4
class ViVoiceDataset(Dataset):
    def __init__(self, data_list, sample_rate=16000, train_mode=True, augmentor=None, aug_dict=None):
        self.data_list = data_list 
        self.sample_rate = sample_rate
        self.target_len = int(sample_rate * 3.0) # 3 giây
        self.train_mode = train_mode
        self.augmentor = augmentor
        self.aug_dict = aug_dict or {}

    def __len__(self):
        return len(self.data_list)

    def _pad_or_crop(self, waveform):
        current_len = waveform.size(1)
        if current_len < self.target_len:
            pad_amt = self.target_len - current_len
            waveform = F.pad(waveform, (0, pad_amt), "constant", 0)
        elif current_len > self.target_len:
            start = np.random.randint(0, current_len - self.target_len)
            waveform = waveform[:, start:start + self.target_len]
        return waveform

    def __getitem__(self, idx):
        if self.train_mode:
            # Train Mode: data_list là list of (wav_path, label)
            wav_path, label = self.data_list[idx]
            waveform, sr = torchaudio.load(wav_path)
            
            if sr != self.sample_rate:
                waveform = torchaudio.transforms.Resample(sr, self.sample_rate)(waveform)
            
            # Random crop 3s (hoặc pad nếu ngắn hơn)
            waveform = self._pad_or_crop(waveform)
            
            waveform = waveform * (1 << 15)
            if waveform.shape[0] > 1:
                waveform = waveform.mean(dim=0, keepdim=True)
                
            waveform_np = waveform[0].numpy()
            if self.augmentor and wav_path in self.aug_dict:
                aug_type = self.aug_dict[wav_path]
                waveform_np = self.augmentor.apply_augmentation(waveform_np, aug_type)

            waveform = torch.FloatTensor(waveform_np).unsqueeze(0)
            fbank = compute_fbank(waveform, sample_rate=self.sample_rate)
            return fbank, label

        else:
            # Val Mode: data_list là list các wav_path từ trials.txt (chỉ cần lấy path)
            wav_path = self.data_list[idx]
            waveform, sr = torchaudio.load(wav_path)
            if sr != self.sample_rate:
                waveform = torchaudio.transforms.Resample(sr, self.sample_rate)(waveform)
                
            if waveform.shape[0] > 1:
                waveform = waveform.mean(dim=0, keepdim=True)
            waveform = waveform * (1 << 15)
            
            # Cắt audio thành nhiều đoạn 3s, overlap 50% (1.5s step)
            step_len = self.target_len // 2
            chunks = []
            
            if waveform.size(1) < self.target_len:
                pad_amt = self.target_len - waveform.size(1)
                padded_wav = F.pad(waveform, (0, pad_amt), "constant", 0)
                chunks.append(padded_wav)
            else:
                for start in range(0, waveform.size(1) - self.target_len + 1, step_len):
                    chunks.append(waveform[:, start:start + self.target_len])
                # Xử lý đoạn lẻ cuối cùng nếu có
                if (waveform.size(1) - self.target_len) % step_len != 0:
                    chunks.append(waveform[:, -self.target_len:])
            
            fbanks = [compute_fbank(chunk, sample_rate=self.sample_rate) for chunk in chunks]
            fbanks_tensor = torch.stack(fbanks) # Shape: [num_chunks, time, freq]
            return wav_path, fbanks_tensor
```

```python
#Cell 5
class SpeakerBalancedBatchSampler(Sampler):
    def __init__(self, records, speakers_per_batch, utts_per_speaker):
        self.speakers_per_batch = speakers_per_batch
        self.utts_per_speaker = utts_per_speaker
        self.spk_to_indices = defaultdict(list)
        for idx, (_, spk_id) in enumerate(records):
            self.spk_to_indices[spk_id].append(idx)
        self.speakers = sorted(self.spk_to_indices.keys())
        self.batch_size = speakers_per_batch * utts_per_speaker
        self.num_batches = max(1, math.ceil(len(records) / self.batch_size))
        self.epoch = 0
        
    def set_epoch(self, epoch): 
        self.epoch = epoch
        
    def __len__(self): 
        return self.num_batches
        
    def __iter__(self):
        rng = random.Random(2025 + self.epoch)
        for _ in range(self.num_batches):
            chosen = rng.sample(self.speakers, min(self.speakers_per_batch, len(self.speakers))) if len(self.speakers) >= self.speakers_per_batch else [rng.choice(self.speakers) for _ in range(self.speakers_per_batch)]
            batch = []
            for spk in chosen:
                idxs = self.spk_to_indices[spk]
                batch.extend(rng.sample(idxs, self.utts_per_speaker) if len(idxs) >= self.utts_per_speaker else rng.choices(idxs, k=self.utts_per_speaker))
            rng.shuffle(batch)
            yield batch

class ViVoiceDataModule(pl.LightningDataModule):
    def __init__(self, data_root, speakers_per_batch=32, utts_per_speaker=2, num_workers=4, aug_ratio=0.5, noise_dir=None): 
        super().__init__()
        self.train_dir  = os.path.join(data_root, "train")
        self.val_dir    = os.path.join(data_root, "val")
        self.speakers_per_batch = speakers_per_batch
        self.utts_per_speaker = utts_per_speaker
        self.num_workers= num_workers
        self.aug_ratio  = aug_ratio
        self.noise_dir  = noise_dir
        
    def setup(self, stage=None):
        if hasattr(self, 'train_list') and self.train_list: return
        
        # Thêm debug kiểm tra thư mục
        print(f"Đang tìm kiếm thư mục train tại: {self.train_dir}")
        if not os.path.exists(self.train_dir):
            raise FileNotFoundError(f"[LỖI] Không tìm thấy thư mục {self.train_dir}. Vui lòng kiểm tra lại DATA_ROOT.")

        speaker_names = sorted([d for d in os.listdir(self.train_dir) if os.path.isdir(os.path.join(self.train_dir, d))])
        print(f"Tìm thấy {len(speaker_names)} speakers trong tập train.")
        
        if len(speaker_names) == 0:
            raise ValueError("[LỖI] Thư mục train trống rỗng hoặc không chứa các thư mục speaker.")

        self.spk2id   = {spk: i for i, spk in enumerate(speaker_names)}
        self.num_classes = len(self.spk2id)
        
        self.train_list = []
        all_train_files = []
        
        print("Quét độ dài file Train để phân bổ Dynamic Capacity...")
        import soundfile as sf
        for spk_name in speaker_names:
            spk_path = os.path.join(self.train_dir, spk_name)
            label = self.spk2id[spk_name]
            wav_files = glob.glob(os.path.join(spk_path, "**/*.wav"), recursive=True)
            
            for wav_path in wav_files:
                try:
                    # Thay thế torchaudio.info bằng soundfile
                    info = sf.info(wav_path)
                    len_sec = info.duration
                    
                    # Tính toán số lần file này được lấy sample trong 1 epoch
                    capacity = max(1, math.ceil(len_sec / 3.0))
                    for _ in range(capacity):
                        self.train_list.append((wav_path, label))
                    all_train_files.append(wav_path)
                except Exception as e: 
                    # Tạm tắt print để tránh lỗi IOPub data rate exceeded trên Jupyter
                    # print(f"Lỗi khi đọc file {wav_path}: {e}") 
                    continue

        if len(self.train_list) == 0:
            raise ValueError("[LỖI] Đã tìm thấy speaker nhưng không tìm thấy file .wav hợp lệ nào bên trong.")

        self.train_aug_dict = self._prepare_augmentation_dict(all_train_files, self.aug_ratio)
        
        # TỰ ĐỘNG SINH TRIALS CHO TẬP VAL
        val_wav_dir = os.path.join(self.val_dir, "wav")
        print(f"Đang tự động sinh cặp trials từ: {val_wav_dir}")
        
        spk_to_wavs = defaultdict(list)
        for spk_id in os.listdir(val_wav_dir):
            spk_path = os.path.join(val_wav_dir, spk_id)
            if os.path.isdir(spk_path):
                # Lấy tất cả file wav của speaker này
                wavs = glob.glob(os.path.join(spk_path, "**/*.wav"), recursive=True)
                if len(wavs) > 0:
                    spk_to_wavs[spk_id] = wavs

        speakers = list(spk_to_wavs.keys())
        positive_pairs = []
        negative_pairs = []

        # 1. Sinh Positive Pairs (Cùng Speaker)
        for spk in speakers:
            wavs = spk_to_wavs[spk]
            if len(wavs) >= 2:
                # Tạo tất cả các cặp có thể (hoặc random sample nếu quá nhiều)
                pairs = list(itertools.combinations(wavs, 2))
                random.shuffle(pairs)
                # Giới hạn max 20 cặp mỗi speaker để tránh lệch dữ liệu
                positive_pairs.extend([(1, p[0], p[1]) for p in pairs[:20]])

        # 2. Sinh Negative Pairs (Khác Speaker)
        num_neg_needed = len(positive_pairs) # Cân bằng số lượng âm/dương
        while len(negative_pairs) < num_neg_needed:
            spk1, spk2 = random.sample(speakers, 2)
            wav1 = random.choice(spk_to_wavs[spk1])
            wav2 = random.choice(spk_to_wavs[spk2])
            negative_pairs.append((0, wav1, wav2))

        # Gộp và shuffle
        self.trials = positive_pairs + negative_pairs
        random.shuffle(self.trials)
        
        unique_wavs = set()
        for label, p1, p2 in self.trials:
            unique_wavs.add(p1)
            unique_wavs.add(p2)
            
        self.valid_list = list(unique_wavs)
        print(f"Đã sinh {len(positive_pairs)} Pos pairs, {len(negative_pairs)} Neg pairs.")
        print(f"Tổng Unique Valid Files cần trích xuất: {len(self.valid_list)}")

    def _prepare_augmentation_dict(self, file_paths, ratio): 
        unique_paths = list(set(file_paths))
        n_aug = int(len(unique_paths) * ratio)
        indices = np.random.permutation(len(unique_paths))
        aug_dict = {}
        # Loại bỏ 'rir_reverb' khỏi danh sách này
        aug_types = ['musan_noise', 'device']
        for i in indices[:n_aug]:
            aug_dict[unique_paths[i]] = np.random.choice(aug_types)
        return aug_dict

    def train_dataloader(self):
        augmentor = SoftAugmentor(noise_dir=self.noise_dir)
        dataset   = ViVoiceDataset(self.train_list, train_mode=True, augmentor=augmentor, aug_dict=self.train_aug_dict)
        batch_sampler = SpeakerBalancedBatchSampler(self.train_list, self.speakers_per_batch, self.utts_per_speaker)
        return DataLoader(dataset, batch_sampler=batch_sampler, num_workers=self.num_workers)
    
    def val_dataloader(self):
        # Val batch size = 1 vì mỗi file có số lượng chunk khác nhau
        dataset = ViVoiceDataset(self.valid_list, train_mode=False)
        return DataLoader(dataset, batch_size=1, shuffle=False, num_workers=self.num_workers)
```

```python
#cell 6
class LitSpeakerModel(pl.LightningModule):
    def __init__(self, pretrained_path, num_classes, lr_backbone=1e-5, lr_head=1e-3, freeze_epochs=2):
        super().__init__()
        self.save_hyperparameters()
        self.best_val_eer = 1.0
        self.backbone, self.config = load_model_pt(pretrained_path)
        embed_dim = self.config['model_args']['embed_dim']

        # ÉP HỌC KỸ 1: Giảm số epoch freeze xuống (ví dụ 1) để backbone được finetune sớm hơn
        for param in self.backbone.parameters():
            param.requires_grad = False
            
        # ÉP HỌC KỸ 2: Tăng margin từ 0.2 lên 0.3 để siết chặt ranh giới giữa các speaker
        self.projection = ArcMarginProjection(in_features=embed_dim, out_features=num_classes, margin=0.3, scale=32.0)
        self.criterion  = torch.nn.CrossEntropyLoss()
        self.validation_step_outputs = []

    def forward(self, x):
        out = self.backbone(x)
        return out[-1] if isinstance(out, tuple) else out

    def on_train_epoch_start(self):
        try:
            dl = self.trainer.train_dataloader
            if hasattr(dl, "batch_sampler") and hasattr(dl.batch_sampler, "set_epoch"):
                dl.batch_sampler.set_epoch(self.current_epoch)
        except Exception:
            pass

        if self.current_epoch == self.hparams.freeze_epochs:
            print(f"\n[INFO] Unfreezing backbone at epoch {self.current_epoch}!")
            for param in self.backbone.parameters():
                param.requires_grad = True

    def training_step(self, batch, batch_idx):
        feats, labels = batch 
        embeds = self(feats)
        logits = self.projection(embeds, labels)
        loss   = self.criterion(logits, labels)
        self.log('train_loss', loss, on_step=True, on_epoch=True, prog_bar=True)
        return loss
    
    def validation_step(self, batch, batch_idx):
        # Batch_size = 1 -> Lấy phần tử đầu tiên
        wav_path = batch[0][0] 
        fbanks = batch[1][0] # Shape: [num_chunks, time, freq]
        
        embeds = self(fbanks) 
        embeds = F.normalize(embeds, p=2, dim=1) # L2 normalize từng segment
        
        avg_embed = torch.mean(embeds, dim=0) # Average embedding
        final_embed = F.normalize(avg_embed, p=2, dim=0) # L2 normalize lần 2
        
        self.validation_step_outputs.append({
            "wav_path": wav_path,
            "embed": final_embed.detach().cpu()
        })
        return final_embed
    
    def on_validation_epoch_end(self):
        if not self.validation_step_outputs: return
        
        emb_dict = {x["wav_path"]: x["embed"] for x in self.validation_step_outputs}
        self.validation_step_outputs.clear()
        
        trials = self.trainer.datamodule.trials
        
        y_true, y_scores = [], []
        for label, path1, path2 in trials:
            # path1, path2 giờ đây đã là đường dẫn tuyệt đối từ data module
            if path1 in emb_dict and path2 in emb_dict:
                score = F.cosine_similarity(emb_dict[path1], emb_dict[path2], dim=0)
                y_scores.append(score.item())
                y_true.append(float(label)) # Đổi sang float để tính loss
                
        if len(y_true) < 2: return
            
        y_true_tensor = torch.tensor(y_true)
        y_scores_tensor = torch.tensor(y_scores)
        
        # 1. TÍNH VAL LOSS MỚI (Dựa trên Cosine similarity)
        # Cosine [-1, 1] -> Đưa về [0, 1] để khớp với nhãn 0/1
        scaled_scores = (y_scores_tensor + 1.0) / 2.0
        val_loss = F.mse_loss(scaled_scores, y_true_tensor)
        self.log('val_loss', val_loss, prog_bar=True)
            
        # 2. TÍNH EER
        fpr, tpr, thresholds = roc_curve(y_true, y_scores)
        fnr = 1 - tpr
        eer_index = np.nanargmin(np.abs(fnr - fpr))
        eer = (fpr[eer_index] + fnr[eer_index]) / 2.0
        
        # 3. TÍNH CHUẨN NORMALIZED minDCF
        p_target = 0.01
        c_miss = 1.0
        c_fa = 1.0
        dcf = c_miss * fnr * p_target + c_fa * fpr * (1 - p_target)
        c_default = min(c_miss * p_target, c_fa * (1 - p_target))
        mindcf_norm = np.min(dcf) / c_default # CHUẨN HÓA
        
        self.log('val_eer', eer, prog_bar=True)
        self.log('val_mindcf', mindcf_norm, prog_bar=False)
        
        print(f"\nEpoch {self.current_epoch} - Val Loss: {val_loss:.4f} | EER: {eer:.4f} | minDCF: {mindcf_norm:.4f}")
        if eer < self.best_val_eer:
            self.best_val_eer = eer
            print(f"✨ NEW BEST EER: {self.best_val_eer:.4f} ✨")

    def configure_optimizers(self):
        backbone_params = list(self.backbone.parameters())
        head_params     = list(self.projection.parameters())
        
        # Dùng AdamW thay vì Adam để decay trọng số tốt hơn (giúp tránh overfitting khi train lâu)
        optimizer = torch.optim.AdamW([
            {"params": backbone_params, "lr": self.hparams.lr_backbone},
            {"params": head_params,     "lr": self.hparams.lr_head},
        ], weight_decay=1e-4)

        # ÉP HỌC KỸ 3: Đổi sang Cosine Annealing. LR sẽ giảm từ từ theo hình sin, 
        # giúp model tìm được điểm minimum tốt hơn khi bạn không quan tâm thời gian train.
        scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
            optimizer, T_max=self.trainer.max_epochs, eta_min=1e-6
        )
        return [optimizer], [scheduler]
```

```python
#Cell 7
DATA_ROOT       = "/kaggle/input/datasets/tranvannha/vivoice34/VoxVietnamese_dataset"
PRETRAINED_DIR  = "/kaggle/input/datasets/minhtu4n/samresnet34-pretrained-voxblink2"
NOISE_DIR       = "/kaggle/input/datasets/minhtu4n/musan-train-set/musan"      
CHECKPOINT_DIR  = "/kaggle/working"

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

dm = ViVoiceDataModule(
    data_root          = DATA_ROOT,
    speakers_per_batch = 16, # ~64 batch size total for T4 16GB
    utts_per_speaker   = 2,  # Yêu cầu >= 2 sample / speaker
    num_workers        = 4,
    aug_ratio          = 0.5,           
    noise_dir          = NOISE_DIR
)
dm.setup()
model = LitSpeakerModel(
    pretrained_path = PRETRAINED_DIR,
    num_classes     = dm.num_classes,
    lr_backbone     = 5e-5,  # Chỉnh một chút để hội tụ tốt hơn với batch_size >= 64
    lr_head         = 1e-3,
    freeze_epochs   = 3 # Unfreeze toàn bộ model từ epoch thứ 3 trở đi
)

checkpoint_callback = ModelCheckpoint(
    dirpath=CHECKPOINT_DIR, filename="best_vivoice_model", monitor="val_eer", mode="min", save_top_k=1
)
early_stop_callback = EarlyStopping(monitor="val_eer", mode="min", patience=5)
lr_monitor = LearningRateMonitor(logging_interval='epoch')

trainer = pl.Trainer(
    max_epochs=150,
    accelerator='gpu',
    devices=1, 
    callbacks=[checkpoint_callback, early_stop_callback, lr_monitor],
    precision="16-mixed",
    accumulate_grad_batches=2,
    num_sanity_val_steps=0,
    enable_progress_bar=False # <--- THÊM DÒNG NÀY ĐỂ TẮT PROGRESS BAR GÂY LỖI RICH
)

trainer.fit(model, datamodule=dm)

# TRÍCH XUẤT TRỌNG SỐ
best_ckpt = checkpoint_callback.best_model_path
lightning_checkpoint = torch.load(best_ckpt, map_location="cpu")
fine_tuned_state_dict = {k[9:]: v for k, v in lightning_checkpoint['state_dict'].items() if k.startswith('backbone.')}
torch.save(fine_tuned_state_dict, "/kaggle/working/simamresnet34_vivoice_finetuned_OLD.pt")
print("✅ Đã lưu trọng số backbone hoàn chỉnh!")
```
